```python

# ==========================================
# BAB 5.2 - SKEWNESS DAN KURTOSIS
# ==========================================

distribution_analysis = pd.DataFrame()

distribution_analysis["Skewness"] = df[numeric_cols].skew()
distribution_analysis["Kurtosis"] = df[numeric_cols].kurt()

# Interpretasi skewness
def interpret_skew(value):
    if value > 0.5:
        return "Positive Skew"
    elif value < -0.5:
        return "Negative Skew"
    else:
        return "Relatif Simetris"

# Interpretasi kurtosis (excess kurtosis)
def interpret_kurt(value):
    if value > 0:
        return "Leptokurtik"
    elif value < 0:
        return "Platikurtik"
    else:
        return "Mesokurtik"

distribution_analysis["Bentuk Skewness"] = (
    distribution_analysis["Skewness"].apply(interpret_skew)
)

distribution_analysis["Bentuk Kurtosis"] = (
    distribution_analysis["Kurtosis"].apply(interpret_kurt)
)

print("HASIL ANALISIS DISTRIBUSI DATA")
display(distribution_analysis.round(4))

# Mengurutkan fitur berdasarkan skewness tertinggi
print("\n10 FITUR DENGAN SKEWNESS TERTINGGI")
display(
    distribution_analysis
    .sort_values("Skewness", ascending=False)
    .head(10)
    .round(4)
)
```

```python
# ==========================================
# BAB 6: TRANSFORMASI DATA
# Persiapan: Menghapus kolom identitas & 8 fitur konstan (zero-variance)
# ==========================================
from sklearn.preprocessing import StandardScaler, MinMaxScaler

df_bab6 = df.copy()

zero_var_cols = [
    'embedded_file_count', 'average_embedded_file_size',
    'xref_count', 'xref_entries', 'submitform_count',
    'jbig2decode_count', 'trailer_count', 'startxref_count'
]

df_bab6 = df_bab6.drop(columns=['file_path'] + zero_var_cols)

print("Dimensi awal:", df.shape)
print("Dimensi setelah hapus file_path & 8 fitur konstan:", df_bab6.shape)
```

```python
# 6.1 Handling Skewness dengan Log Transformation: log(1 + x)
binary_cols = ['encrypted', 'valid_pdf_header', 'uses_nonstandard_port', 'used_ocr', 'label']
num_cols = [col for col in df_bab6.columns if col not in binary_cols]

# Hitung skewness sebelum transformasi
skew_before = df_bab6[num_cols].skew()
highly_skewed = skew_before[skew_before.abs() > 0.5].index.tolist()

print(f"Jumlah fitur numerik dengan |skewness| > 0.5: {len(highly_skewed)} dari {len(num_cols)} fitur")

# Terapkan np.log1p (log(x + 1)) karena banyak nilai 0
df_log = df_bab6.copy()
for col in highly_skewed:
    df_log[col] = np.log1p(df_log[col])

# Perbandingan skewness sebelum & sesudah
skew_after = df_log[highly_skewed].skew()
skew_comp = pd.DataFrame({
    'Skewness_Sebelum': skew_before[highly_skewed],
    'Skewness_Sesudah': skew_after,
    'Selisih_Penurunan': skew_before[highly_skewed] - skew_after
}).round(4)

print("\nPerbandingan Skewness (Diurutkan dari Skewness Awal Tertinggi):")
display(skew_comp.sort_values(by='Skewness_Sebelum', ascending=False))

# Visualisasi Perbandingan Sebelum vs Sesudah pada 4 fitur struktural utama
fig, axes = plt.subplots(4, 2, figsize=(12, 14))
plot_cols = ['file_size', 'page_count', 'object_count', 'text_length']

for i, col in enumerate(plot_cols):
    sns.histplot(df_bab6[col], bins=40, kde=True, ax=axes[i, 0], color='steelblue')
    axes[i, 0].set_title(f'Sebelum Log1p: {col} (Skew = {skew_before[col]:.2f})')

    sns.histplot(df_log[col], bins=40, kde=True, ax=axes[i, 1], color='darkorange')
    axes[i, 1].set_title(f'Sesudah Log1p: {col} (Skew = {skew_after[col]:.2f})')

plt.tight_layout()
plt.show()
```

```python
# 6.2 StandardScaler vs MinMaxScaler
feature_cols = [col for col in df_log.columns if col != 'label']

# 1. Standardisasi (Mean = 0, Std = 1)
std_scaler = StandardScaler()
df_std = pd.DataFrame(std_scaler.fit_transform(df_log[feature_cols]), columns=feature_cols)

# 2. Normalisasi (Rentang 0 - 1)
mm_scaler = MinMaxScaler()
df_minmax = pd.DataFrame(mm_scaler.fit_transform(df_log[feature_cols]), columns=feature_cols)

sample_scale_cols = ['file_size', 'page_count', 'object_count', 'text_length', 'stream_count', 'entropy_of_streams']

print("=== SEBELUM SCALING (DATA ASLI) ===")
display(df_bab6[sample_scale_cols].describe().T[['mean', 'std', 'min', 'max']].round(4))

print("\n=== SESUDAH STANDARD SCALER (pada data log1p) ===")
display(df_std[sample_scale_cols].describe().T[['mean', 'std', 'min', 'max']].round(4))

print("\n=== SESUDAH MINMAX SCALER (pada data log1p) ===")
display(df_minmax[sample_scale_cols].describe().T[['mean', 'std', 'min', 'max']].round(4))
```

```python
# 6.4 Pembuatan Fitur Baru (Rasio, Agregasi, dan Binning)
df_fe = df_bab6.copy()

# 1. Fitur Rasio (+1 pada penyebut agar tidak dibagi 0)
df_fe['bytes_per_page'] = df_fe['file_size'] / (df_fe['page_count'] + 1)
df_fe['objects_per_page'] = df_fe['object_count'] / (df_fe['page_count'] + 1)
df_fe['streams_per_object'] = df_fe['stream_count'] / (df_fe['object_count'] + 1)

# 2. Fitur Agregasi Perilaku Mencurigakan
df_fe['total_suspicious_actions'] = (
    df_fe['js_count'] + df_fe['javascript_count'] +
    df_fe['openaction_count'] + df_fe['launch_count'] +
    df_fe['aa_count'] + df_fe['action_count']
)
df_fe['has_javascript'] = ((df_fe['js_count'] > 0) | (df_fe['javascript_count'] > 0)).astype(int)

# 3. Penyederhanaan Bins (Kategori Ukuran File berdasarkan kuartil)
df_fe['file_size_category'] = pd.qcut(
    df_fe['file_size'], q=3, labels=['Kecil', 'Sedang', 'Besar']
)

# Contoh penerapan Ordinal Encoding pada fitur kategorikal baru 'file_size_category'
ordinal_map = {'Kecil': 0, 'Sedang': 1, 'Besar': 2}
df_fe['file_size_category_encoded'] = df_fe['file_size_category'].map(ordinal_map)

# Tampilkan perbandingan rata-rata fitur baru berdasarkan label
new_num_cols = ['bytes_per_page', 'objects_per_page', 'streams_per_object', 'total_suspicious_actions', 'has_javascript']
print("=== RATA-RATA FITUR BARU BERDASARKAN LABEL ===")
display(df_fe.groupby('label')[new_num_cols].mean().T.rename(columns={0: 'Benign', 1: 'Malicious'}).round(4))

print("\n=== DISTRIBUSI KATEGORI UKURAN FILE (BINNING) PER LABEL (%) ===")
ct = pd.crosstab(df_fe['file_size_category'], df_fe['label'], normalize='columns') * 100
ct.columns = ['Benign (%)', 'Malicious (%)']
display(ct.round(2))
```

```python
# ==========================================
# BAB 8: PENANGANAN PENCILAN (HANDLING OUTLIERS)
# 8.1 Deteksi Pencilan (IQR & Z-Score)
# ==========================================
from scipy import stats

outlier_cols = [
    'file_size', 'page_count', 'object_count', 'text_length',
    'stream_count', 'metadata_size', 'total_filters',
    'uri_count', 'javascript_count', 'js_count'
]

outlier_summary = []

for col in outlier_cols:
    # 1. Metode IQR
    Q1 = df_bab6[col].quantile(0.25)
    Q3 = df_bab6[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    iqr_count = ((df_bab6[col] < lower_bound) | (df_bab6[col] > upper_bound)).sum()

    # 2. Metode Z-Score (|Z| > 3)
    z_scores = np.abs(stats.zscore(df_bab6[col]))
    z_count = (z_scores > 3).sum()

    outlier_summary.append({
        'Fitur': col,
        'Q1': round(Q1, 2),
        'Q3': round(Q3, 2),
        'Batas_Atas_IQR': round(upper_bound, 2),
        'Outlier_IQR': iqr_count,
        'Persen_IQR (%)': round((iqr_count / len(df_bab6)) * 100, 2),
        'Outlier_ZScore': z_count,
        'Persen_ZScore (%)': round((z_count / len(df_bab6)) * 100, 2)
    })

outlier_df = pd.DataFrame(outlier_summary)
print("=== RINGKASAN DETEKSI PENCILAN (IQR vs Z-SCORE) ===")
display(outlier_df)
```

```python
# Visualisasi Boxplot Sebelum Penanganan pada 4 fitur utama (Data Asli)
fig, axes = plt.subplots(2, 2, figsize=(14, 8))
plot_out_cols = ['file_size', 'page_count', 'object_count', 'text_length']

for ax, col in zip(axes.flatten(), plot_out_cols):
    sns.boxplot(x=df_bab6[col], ax=ax, color='lightcoral')
    ax.set_title(f'Boxplot Sebelum Penanganan: {col}')

plt.tight_layout()
plt.show()

# Cek proporsi Benign vs Malicious pada pencilan ekstrem (untuk justifikasi domain knowledge)
print("=== PROPORSI LABEL PADA DATA PENCILAN (METODE IQR) ===")
domain_check = []
for col in ['file_size', 'object_count', 'javascript_count', 'js_count', 'uri_count']:
    Q1 = df_bab6[col].quantile(0.25)
    Q3 = df_bab6[col].quantile(0.75)
    upper = Q3 + 1.5 * (Q3 - Q1)
    out_subset = df_bab6[df_bab6[col] > upper]

    benign_n = (out_subset['label'] == 0).sum()
    malicious_n = (out_subset['label'] == 1).sum()
    total_n = len(out_subset)

    domain_check.append({
        'Fitur': col,
        'Total_Outlier_IQR': total_n,
        'Benign (0)': benign_n,
        'Malicious (1)': malicious_n,
        'Persen_Malicious (%)': round((malicious_n / total_n) * 100, 2) if total_n > 0 else 0
    })

display(pd.DataFrame(domain_check))
```
